# Decision Tree Based ID3 Algorithm

This notebook demonstrates the implementation of the ID3 (Iterative Dichotomiser 3) algorithm for building a decision tree classifier from training data and using it to classify new samples.

## Importing Required Libraries

We'll use pandas for data manipulation, numpy for numerical operations, and sklearn for implementing the ID3 algorithm and evaluating our model.

```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier, export_text, export_graphviz
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
import matplotlib.pyplot as plt
from sklearn import tree
import graphviz

# For reproducibility
np.random.seed(42)
```

## Creating Sample Data

For this demonstration, we'll create a sample dataset for a simple classification problem. Alternatively, you can load your own dataset from a CSV file.

```python
# Uncomment below code to read data from a CSV file
# data = pd.read_csv('your_data.csv')

# For this demonstration, let's create a simple dataset
# This represents a play tennis decision based on weather conditions
data = {
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain', 'Rain', 'Overcast', 'Sunny', 'Sunny', 'Rain', 'Sunny', 'Overcast', 'Overcast', 'Rain'],
    'Temperature': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool', 'Cool', 'Cool', 'Mild', 'Cool', 'Mild', 'Mild', 'Mild', 'Hot', 'Mild'],
    'Humidity': ['High', 'High', 'High', 'High', 'Normal', 'Normal', 'Normal', 'High', 'Normal', 'Normal', 'Normal', 'High', 'Normal', 'High'],
    'Windy': ['False', 'True', 'False', 'False', 'False', 'True', 'True', 'False', 'False', 'False', 'True', 'True', 'False', 'True'],
    'Play': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes', 'Yes', 'Yes', 'Yes', 'Yes', 'No']
}

df = pd.DataFrame(data)
df.head()
```

```python
# Save this sample dataset to a CSV file for future use
df.to_csv('play_tennis_data.csv', index=False)
print("Data saved to 'play_tennis_data.csv'")
```

## Data Preprocessing

We need to convert categorical variables to numeric form for the decision tree algorithm.

```python
# Convert categorical variables to numeric using one-hot encoding
df_encoded = pd.get_dummies(df, columns=['Outlook', 'Temperature', 'Humidity', 'Windy'], drop_first=False)

# Prepare features and target variables
X = df_encoded.drop('Play', axis=1)
y = df_encoded['Play'].map({'Yes': 1, 'No': 0})

# Display the encoded data
print("Feature columns after encoding:")
print(X.columns.tolist())
X.head()
```

## Building the ID3 Decision Tree

We'll use scikit-learn's DecisionTreeClassifier with entropy criterion (which is the basis of the ID3 algorithm).

```python
# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Create ID3 Decision Tree classifier
# Using entropy as the criterion which is the core of ID3 algorithm
id3_tree = DecisionTreeClassifier(criterion='entropy', random_state=42)

# Train the model
id3_tree.fit(X_train, y_train)

# Make predictions on the test set
y_pred = id3_tree.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy:.4f}")
print("Classification Report:")
print(report)
```

## Visualizing the Decision Tree

Now let's visualize the constructed decision tree to better understand the classification rules.

```python
# Visualize the decision tree as text
tree_text = export_text(id3_tree, feature_names=list(X.columns))
print("Decision Tree Structure:")
print(tree_text)
```

```python
# Visualize the decision tree graphically
plt.figure(figsize=(20, 10))
tree.plot_tree(id3_tree, feature_names=list(X.columns), 
               class_names=['No', 'Yes'], filled=True, rounded=True)
plt.title("ID3 Decision Tree for Tennis Play Dataset")
plt.show()
```

## Classifying New Samples

Now that we have our trained model, let's classify a new sample.

```python
# Create a function to classify new samples
def classify_new_sample(outlook, temperature, humidity, windy):
    # Create a dataframe with the new sample
    new_sample = pd.DataFrame({
        'Outlook': [outlook],
        'Temperature': [temperature],
        'Humidity': [humidity],
        'Windy': [windy]
    })
    
    # Encode the new sample using one-hot encoding
    # We need to ensure it has the same columns as our training data
    new_sample_encoded = pd.get_dummies(new_sample, columns=['Outlook', 'Temperature', 'Humidity', 'Windy'], drop_first=False)
    
    # Add missing columns if any
    for col in X.columns:
        if col not in new_sample_encoded.columns:
            new_sample_encoded[col] = 0
            
    # Ensure columns are in the same order as the training data
    new_sample_encoded = new_sample_encoded[X.columns]
    
    # Make a prediction
    prediction = id3_tree.predict(new_sample_encoded)
    probability = id3_tree.predict_proba(new_sample_encoded)
    
    result = "Yes" if prediction[0] == 1 else "No"
    
    return result, probability

# Try with a new sample
outlook = "Sunny"
temperature = "Mild"
humidity = "Normal"
windy = "False"

result, probability = classify_new_sample(outlook, temperature, humidity, windy)
print(f"Should we play tennis with Outlook={outlook}, Temperature={temperature}, Humidity={humidity}, Windy={windy}?")
print(f"Prediction: {result}")
print(f"Probability: No={probability[0][0]:.4f}, Yes={probability[0][1]:.4f}")
```
