## 1. Calculate Important Numbers from Small Dataset

In this section, we'll calculate statistical measures and create functions to predict outcomes based on a small dataset.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Sample small dataset
data = [12, 15, 18, 22, 30, 31, 35, 38, 42, 48]

# Calculate statistical measures
mean_value = np.mean(data)
median_value = np.median(data)
mode_value = stats.mode(data)[0][0]
std_dev = np.std(data)
variance = np.var(data)
min_value = np.min(data)
max_value = np.max(data)
range_value = max_value - min_value

# Print the statistical measures
print(f"Dataset: {data}")
print(f"Mean: {mean_value}")
print(f"Median: {median_value}")
print(f"Mode: {mode_value}")
print(f"Standard Deviation: {std_dev}")
print(f"Variance: {variance}")
print(f"Range: {range_value}")
print(f"Min: {min_value}")
print(f"Max: {max_value}")
```

### Prediction Functions Based on Dataset

```python
# Function to predict if a new value is above or below the mean
def predict_relative_to_mean(value, data):
    mean_val = np.mean(data)
    if value > mean_val:
        return f"{value} is above the mean ({mean_val})"
    elif value < mean_val:
        return f"{value} is below the mean ({mean_val})"
    else:
        return f"{value} is exactly the mean"

# Function to predict which quartile a value falls into
def predict_quartile(value, data):
    q1 = np.percentile(data, 25)
    q2 = np.percentile(data, 50)
    q3 = np.percentile(data, 75)
    
    if value < q1:
        return f"{value} is in the 1st quartile (below {q1})"
    elif value < q2:
        return f"{value} is in the 2nd quartile (between {q1} and {q2})"
    elif value < q3:
        return f"{value} is in the 3rd quartile (between {q2} and {q3})"
    else:
        return f"{value} is in the 4th quartile (above {q3})"

# Function to predict z-score and interpret it
def predict_zscore(value, data):
    mean_val = np.mean(data)
    std_dev = np.std(data)
    z_score = (value - mean_val) / std_dev
    
    if abs(z_score) <= 1:
        interpretation = "within 1 standard deviation (common)"
    elif abs(z_score) <= 2:
        interpretation = "within 2 standard deviations (somewhat uncommon)"
    elif abs(z_score) <= 3:
        interpretation = "within 3 standard deviations (rare)"
    else:
        interpretation = "beyond 3 standard deviations (very rare/outlier)"
    
    return f"{value} has a z-score of {z_score:.2f} ({interpretation})"

# Test predictions with some values
test_values = [20, 40, 60]
for val in test_values:
    print(predict_relative_to_mean(val, data))
    print(predict_quartile(val, data))
    print(predict_zscore(val, data))
    print()
```

## 3. Build an Artificial Neural Network with Backpropagation

In this section, we'll implement a simple Artificial Neural Network using Python and demonstrate the backpropagation algorithm.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix
import seaborn as sns

# Define the Neural Network class
class NeuralNetwork:
    def __init__(self, input_size, hidden_size, output_size, learning_rate=0.01):
        # Initialize parameters
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size
        self.learning_rate = learning_rate
        
        # Initialize weights and biases
        self.W1 = np.random.randn(self.input_size, self.hidden_size) * 0.01
        self.b1 = np.zeros((1, self.hidden_size))
        self.W2 = np.random.randn(self.hidden_size, self.output_size) * 0.01
        self.b2 = np.zeros((1, self.output_size))
        
        # Initialize arrays to store parameters for future visualization
        self.loss_history = []
        self.accuracy_history = []
        
    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))
    
    def sigmoid_derivative(self, x):
        return x * (1 - x)
    
    def forward(self, X):
        # Forward propagation
        self.z1 = np.dot(X, self.W1) + self.b1
        self.a1 = self.sigmoid(self.z1)
        self.z2 = np.dot(self.a1, self.W2) + self.b2
        self.a2 = self.sigmoid(self.z2)
        return self.a2
    
    def backward(self, X, y, output):
        # Backpropagation
        self.error = y - output
        self.delta_output = self.error * self.sigmoid_derivative(output)
        
        self.error_hidden = self.delta_output.dot(self.W2.T)
        self.delta_hidden = self.error_hidden * self.sigmoid_derivative(self.a1)
        
        # Update weights and biases
        self.W2 += self.a1.T.dot(self.delta_output) * self.learning_rate
        self.b2 += np.sum(self.delta_output, axis=0, keepdims=True) * self.learning_rate
        self.W1 += X.T.dot(self.delta_hidden) * self.learning_rate
        self.b1 += np.sum(self.delta_hidden, axis=0, keepdims=True) * self.learning_rate
    
    def train(self, X, y, epochs=10000):
        for epoch in range(epochs):
            # Forward propagation
            output = self.forward(X)
            
            # Backpropagation
            self.backward(X, y, output)
            
            # Calculate and store loss
            loss = np.mean(np.square(y - output))
            self.loss_history.append(loss)
            
            # Calculate and store accuracy
            predictions = (output > 0.5).astype(int)
            accuracy = accuracy_score(y, predictions)
            self.accuracy_history.append(accuracy)
            
            # Print status every 1000 epochs
            if epoch % 1000 == 0:
                print(f"Epoch {epoch}: Loss = {loss:.6f}, Accuracy = {accuracy:.2f}")
    
    def predict(self, X):
        output = self.forward(X)
        return (output > 0.5).astype(int)

# Generate a simple binary classification dataset
X, y = make_classification(n_samples=200, n_features=2, n_informative=2, 
                           n_redundant=0, n_clusters_per_class=1, random_state=42)

# Reshape y to be a column vector
y = y.reshape(-1, 1)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Standardize the data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Create and train the neural network
nn = NeuralNetwork(input_size=2, hidden_size=4, output_size=1, learning_rate=0.1)
nn.train(X_train_scaled, y_train, epochs=5000)

# Make predictions
y_pred = nn.predict(X_test_scaled)

# Calculate and print the final accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"\nFinal Test Accuracy: {accuracy:.4f}")

# Plot the decision boundary
plt.figure(figsize=(14, 5))

# Plot training progress
plt.subplot(1, 3, 1)
plt.plot(nn.loss_history)
plt.title('Loss over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')

plt.subplot(1, 3, 2)
plt.plot(nn.accuracy_history)
plt.title('Accuracy over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')

# Plot the confusion matrix
plt.subplot(1, 3, 3)
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title(f'Confusion Matrix\nAccuracy: {accuracy:.4f}')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')

plt.tight_layout()
plt.show()

# Visualize the decision boundary
plt.figure(figsize=(10, 8))

# Define the mesh grid
h = 0.02  # step size in the mesh
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))

# Scale the mesh grid and predict
mesh_points = np.c_[xx.ravel(), yy.ravel()]
mesh_points_scaled = scaler.transform(mesh_points)
Z = nn.predict(mesh_points_scaled)
Z = Z.reshape(xx.shape)

# Plot the decision boundary
plt.contourf(xx, yy, Z, cmap=plt.cm.Spectral, alpha=0.8)

# Plot the training points
plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train.flatten(), cmap=plt.cm.Spectral, edgecolors='k')
plt.title('Neural Network Decision Boundary with Training Data')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

plt.show()
```

## 4. Apply Bayes' Rule for Conditional Probability

In this section, we'll use Python to calculate the probability of a student being absent given that it is Friday using Bayes' theorem.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Let's create some sample data for student attendance
# We'll simulate 200 days of attendance records
np.random.seed(42)

# Generate days of the week (0: Monday, 1: Tuesday, ..., 4: Friday)
days = np.random.choice(range(5), size=200)

# Set up probabilities for absence based on the day
# Students are more likely to be absent on Friday
absence_probabilities = {0: 0.05, 1: 0.08, 2: 0.10, 3: 0.12, 4: 0.25}

# Generate absence data
absences = []
for day in days:
    if np.random.random() < absence_probabilities[day]:
        absences.append(1)  # Absent
    else:
        absences.append(0)  # Present

# Create a dataframe with the data
data = pd.DataFrame({'day': days, 'absent': absences})
data['day_name'] = data['day'].map({0: 'Monday', 1: 'Tuesday', 2: 'Wednesday', 
                                    3: 'Thursday', 4: 'Friday'})

print("Sample of the attendance data:")
print(data.head(10))

# Calculate the probability of being absent on each day
absence_by_day = data.groupby('day_name')['absent'].mean()
print("\nProbability of absence by day:")
print(absence_by_day)

# Calculate the probabilities needed for Bayes' theorem
# P(Absent) - Prior probability of absence
p_absent = data['absent'].mean()

# P(Friday) - Prior probability of it being Friday
p_friday = (data['day'] == 4).mean()

# P(Friday | Absent) - Likelihood: Probability of it being Friday given the student is absent
p_friday_given_absent = data[data['absent'] == 1]['day'].apply(lambda x: x == 4).mean()

# Using Bayes' theorem to calculate P(Absent | Friday) - Posterior probability
# P(A|B) = P(B|A) * P(A) / P(B)
p_absent_given_friday = (p_friday_given_absent * p_absent) / p_friday

print(f"\nBayes' Theorem Calculation:")
print(f"P(Absent) = {p_absent:.4f}")
print(f"P(Friday) = {p_friday:.4f}")
print(f"P(Friday | Absent) = {p_friday_given_absent:.4f}")
print(f"P(Absent | Friday) = {p_absent_given_friday:.4f}")

# Verify the calculation directly from the data
direct_calculation = data[data['day'] == 4]['absent'].mean()
print(f"\nDirect calculation of P(Absent | Friday) = {direct_calculation:.4f}")

# Visualize the probabilities
plt.figure(figsize=(12, 6))

# Plot absence probability by day
plt.subplot(1, 2, 1)
absence_by_day.plot(kind='bar', color='skyblue')
plt.title('Probability of Absence by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Probability of Absence')
plt.ylim(0, 0.3)

# Create a visual representation of Bayes' theorem
plt.subplot(1, 2, 2)
labels = ['Overall\nAbsence Rate', 'Absence Rate\non Friday']
values = [p_absent, p_absent_given_friday]
plt.bar(labels, values, color=['blue', 'red'])
plt.title("Bayes' Theorem Results")
plt.ylabel('Probability')
plt.ylim(0, 0.3)

plt.tight_layout()
plt.show()

# Additional analysis: How much more likely are students to be absent on Friday?
relative_risk = p_absent_given_friday / p_absent
print(f"\nStudents are {relative_risk:.2f} times more likely to be absent on Friday compared to the overall absence rate.")
```

## 5. k-Nearest Neighbour Algorithm for Iris Dataset

In this section, we'll implement the k-Nearest Neighbour algorithm to classify the Iris dataset and evaluate its performance.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler
import seaborn as sns
from collections import Counter

# Load the Iris dataset
iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names
target_names = iris.target_names

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Standardize the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Define the KNN algorithm
class KNN:
    def __init__(self, k=3):
        self.k = k
        
    def fit(self, X, y):
        self.X_train = X
        self.y_train = y
        
    def euclidean_distance(self, x1, x2):
        return np.sqrt(np.sum((x1 - x2) ** 2))
    
    def predict(self, X):
        y_pred = []
        for x in X:
            distances = []
            for i, x_train in enumerate(self.X_train):
                dist = self.euclidean_distance(x, x_train)
                distances.append((dist, self.y_train[i]))
            
            # Sort distances and get the k nearest neighbors
            distances.sort(key=lambda x: x[0])
            k_nearest = distances[:self.k]
            
            # Get the most common class among k-nearest neighbors
            k_nearest_labels = [label for _, label in k_nearest]
            most_common = Counter(k_nearest_labels).most_common(1)[0][0]
            y_pred.append(most_common)
        
        return np.array(y_pred)

# Function to evaluate the KNN model with different k values
def evaluate_knn(X_train, y_train, X_test, y_test, k_values):
    accuracy_scores = []
    
    for k in k_values:
        knn = KNN(k=k)
        knn.fit(X_train, y_train)
        y_pred = knn.predict(X_test)
        accuracy = accuracy_score(y_test, y_pred)
        accuracy_scores.append(accuracy)
        print(f"k={k}, Accuracy: {accuracy:.4f}")
    
    return accuracy_scores

# Test different k values
k_values = [1, 3, 5, 7, 9, 11, 13, 15]
print("KNN Accuracy for different k values:")
accuracy_scores = evaluate_knn(X_train_scaled, y_train, X_test_scaled, y_test, k_values)

# Find the best k value
best_k = k_values[np.argmax(accuracy_scores)]
print(f"\nBest k value: {best_k} with accuracy: {max(accuracy_scores):.4f}")

# Create a KNN model with the best k value
knn_best = KNN(k=best_k)
knn_best.fit(X_train_scaled, y_train)
y_pred_best = knn_best.predict(X_test_scaled)

# Evaluate the best model
print("\nClassification Report:")
print(classification_report(y_test, y_pred_best, target_names=target_names))

# Plot the results
plt.figure(figsize=(14, 10))

# Plot accuracy vs k value
plt.subplot(2, 2, 1)
plt.plot(k_values, accuracy_scores, marker='o', linestyle='-', color='blue')
plt.title('Accuracy vs k Value')
plt.xlabel('k Value')
plt.ylabel('Accuracy')
plt.grid(True)

# Plot confusion matrix
plt.subplot(2, 2, 2)
cm = confusion_matrix(y_test, y_pred_best)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=target_names, yticklabels=target_names)
plt.title('Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')

# Visualize the data in 2D (using the first two features)
plt.subplot(2, 2, 3)
for i, target_name in enumerate(target_names):
    plt.scatter(X_test[y_test == i, 0], X_test[y_test == i, 1], label=target_name)
plt.title('Iris Dataset - Actual Classes')
plt.xlabel(feature_names[0])
plt.ylabel(feature_names[1])
plt.legend()

plt.subplot(2, 2, 4)
for i, target_name in enumerate(target_names):
    plt.scatter(X_test[y_pred_best == i, 0], X_test[y_pred_best == i, 1], label=target_name)
plt.title('Iris Dataset - Predicted Classes')
plt.xlabel(feature_names[0])
plt.ylabel(feature_names[1])
plt.legend()

plt.tight_layout()
plt.show()

# Visualize a single prediction example
random_idx = np.random.randint(0, len(X_test))
test_point = X_test_scaled[random_idx]
actual_class = y_test[random_idx]
predicted_class = y_pred_best[random_idx]

# Calculate distances to all training points for visualization
distances = []
for i, x_train in enumerate(X_train_scaled):
    dist = np.sqrt(np.sum((test_point - x_train) ** 2))
    distances.append((dist, y_train[i], i))

# Sort distances and get the k nearest neighbors
distances.sort(key=lambda x: x[0])
k_nearest = distances[:best_k]

plt.figure(figsize=(10, 8))
colors = ['red', 'green', 'blue']

# Plot all training points
for i, class_name in enumerate(target_names):
    plt.scatter(X_train[y_train == i, 0], X_train[y_train == i, 1], 
                color=colors[i], label=class_name, alpha=0.3, s=30)

# Plot the test point
plt.scatter(X_test[random_idx, 0], X_test[random_idx, 1], 
            color='black', marker='*', s=200, 
            label=f'Test Point (True: {target_names[actual_class]}, Pred: {target_names[predicted_class]})')

# Highlight the k-nearest neighbors
for _, label, idx in k_nearest:
    plt.scatter(X_train[idx, 0], X_train[idx, 1], 
                color=colors[label], edgecolors='black', s=100, alpha=0.8)

plt.title(f'KNN Visualization (k={best_k})')
plt.xlabel(feature_names[0])
plt.ylabel(feature_names[1])
plt.legend()
plt.show()
```

## 6. Simple Linear Regression from CSV

In this section, we'll read training data from a CSV file and implement a simple Linear Regression algorithm to predict outcomes.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# First, let's generate a sample CSV file for demonstration purposes
np.random.seed(42)

# Generate synthetic data
X = np.random.uniform(0, 10, 100).reshape(-1, 1)
y = 2 * X.flatten() + 1 + np.random.normal(0, 1, 100)

# Create a dataframe
data = pd.DataFrame({'X': X.flatten(), 'y': y})

# Save to CSV
csv_path = 'linear_regression_data.csv'
data.to_csv(csv_path, index=False)
print(f"Created sample CSV file: {csv_path}")
print(data.head())

# Read the data from CSV
data = pd.read_csv(csv_path)
print("\nData loaded from CSV:")
print(data.head())

# Split into features and target
X = data['X'].values.reshape(-1, 1)
y = data['y'].values

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Implement simple linear regression from scratch
class SimpleLinearRegression:
    def __init__(self):
        self.slope = None
        self.intercept = None
    
    def fit(self, X, y):
        # Calculate means
        x_mean = np.mean(X)
        y_mean = np.mean(y)
        
        # Calculate numerator and denominator for slope
        numerator = 0
        denominator = 0
        
        for i in range(len(X)):
            numerator += (X[i] - x_mean) * (y[i] - y_mean)
            denominator += (X[i] - x_mean) ** 2
        
        # Calculate slope and intercept
        self.slope = numerator / denominator
        self.intercept = y_mean - (self.slope * x_mean)
        
    def predict(self, X):
        return self.slope * X + self.intercept

# Train the model
model = SimpleLinearRegression()
model.fit(X_train.flatten(), y_train)

# Make predictions
y_pred_train = model.predict(X_train.flatten())
y_pred_test = model.predict(X_test.flatten())

# Calculate metrics
train_mse = mean_squared_error(y_train, y_pred_train)
train_r2 = r2_score(y_train, y_pred_train)
test_mse = mean_squared_error(y_test, y_pred_test)
test_r2 = r2_score(y_test, y_pred_test)

# Print results
print(f"\nSimple Linear Regression Results:")
print(f"Slope: {model.slope:.4f}")
print(f"Intercept: {model.intercept:.4f}")
print(f"Training MSE: {train_mse:.4f}")
print(f"Training R^2: {train_r2:.4f}")
print(f"Testing MSE: {test_mse:.4f}")
print(f"Testing R^2: {test_r2:.4f}")

# Visualize the results
plt.figure(figsize=(14, 6))

# Plot training data and regression line
plt.subplot(1, 2, 1)
plt.scatter(X_train, y_train, color='blue', alpha=0.7, label='Training Data')
plt.plot(X_train, y_pred_train, color='red', linewidth=2, label=f'Regression Line: y = {model.slope:.2f}x + {model.intercept:.2f}')
plt.title('Simple Linear Regression - Training Data')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot test data and regression line
plt.subplot(1, 2, 2)
plt.scatter(X_test, y_test, color='green', alpha=0.7, label='Test Data')
plt.plot(X_test, y_pred_test, color='red', linewidth=2, label=f'Regression Line: y = {model.slope:.2f}x + {model.intercept:.2f}')
plt.title('Simple Linear Regression - Test Data')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Visualize residuals
plt.figure(figsize=(14, 6))

# Plot training residuals
plt.subplot(1, 2, 1)
train_residuals = y_train - y_pred_train
plt.scatter(y_pred_train, train_residuals, color='blue', alpha=0.7)
plt.axhline(y=0, color='red', linestyle='-', linewidth=2)
plt.title('Residual Plot - Training Data')
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.grid(True, alpha=0.3)

# Plot test residuals
plt.subplot(1, 2, 2)
test_residuals = y_test - y_pred_test
plt.scatter(y_pred_test, test_residuals, color='green', alpha=0.7)
plt.axhline(y=0, color='red', linestyle='-', linewidth=2)
plt.title('Residual Plot - Test Data')
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 7. Logistic Regression (Binomial) from CSV

In this section, we'll read training data from a CSV file and implement Logistic Regression to classify binary outcomes.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, roc_curve, auc
from sklearn.model_selection import train_test_split
import seaborn as sns

# Generate a sample CSV file for demonstration purposes
np.random.seed(42)

# Generate synthetic binary classification data
n_samples = 100
x1 = np.random.normal(2, 1, int(n_samples/2))
x2 = np.random.normal(4, 1, int(n_samples/2))
X = np.concatenate((x1, x2)).reshape(-1, 1)

# Create labels: 0 for x1, 1 for x2
y = np.concatenate((np.zeros(int(n_samples/2)), np.ones(int(n_samples/2))))

# Add some noise to make it more realistic
X += np.random.normal(0, 0.1, X.shape)

# Create a dataframe
data = pd.DataFrame({'X': X.flatten(), 'y': y})

# Save to CSV
csv_path = 'logistic_regression_data.csv'
data.to_csv(csv_path, index=False)
print(f"Created sample CSV file: {csv_path}")
print(data.head())

# Read the data from CSV
data = pd.read_csv(csv_path)
print("\nData loaded from CSV:")
print(data.head())

# Split into features and target
X = data['X'].values.reshape(-1, 1)
y = data['y'].values

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Implement logistic regression from scratch
class LogisticRegression:
    def __init__(self, learning_rate=0.01, num_iterations=1000):
        self.learning_rate = learning_rate
        self.num_iterations = num_iterations
        self.weights = None
        self.bias = None
        self.cost_history = []
        
    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))
    
    def fit(self, X, y):
        # Initialize parameters
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0
        
        # Gradient descent
        for _ in range(self.num_iterations):
            # Linear combination
            linear_model = np.dot(X, self.weights) + self.bias
            
            # Sigmoid function
            y_predicted = self.sigmoid(linear_model)
            
            # Compute gradients
            dw = (1 / n_samples) * np.dot(X.T, (y_predicted - y))
            db = (1 / n_samples) * np.sum(y_predicted - y)
            
            # Update parameters
            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db
            
            # Calculate cost for tracking
            cost = -np.mean(y * np.log(y_predicted + 1e-15) + (1 - y) * np.log(1 - y_predicted + 1e-15))
            self.cost_history.append(cost)
    
    def predict_proba(self, X):
        linear_model = np.dot(X, self.weights) + self.bias
        return self.sigmoid(linear_model)
    
    def predict(self, X, threshold=0.5):
        probabilities = self.predict_proba(X)
        return (probabilities >= threshold).astype(int)

# Train the model
model = LogisticRegression(learning_rate=0.1, num_iterations=2000)
model.fit(X_train, y_train)

# Make predictions
y_prob_train = model.predict_proba(X_train)
y_pred_train = model.predict(X_train)
y_prob_test = model.predict_proba(X_test)
y_pred_test = model.predict(X_test)

# Calculate metrics
train_accuracy = accuracy_score(y_train, y_pred_train)
test_accuracy = accuracy_score(y_test, y_pred_test)

# Print results
print(f"\nLogistic Regression Results:")
print(f"Weight: {model.weights[0]:.4f}")
print(f"Bias: {model.bias:.4f}")
print(f"Training Accuracy: {train_accuracy:.4f}")
print(f"Testing Accuracy: {test_accuracy:.4f}")

# Print classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred_test))

# Visualize the results
plt.figure(figsize=(16, 12))

# Plot training history
plt.subplot(2, 2, 1)
plt.plot(model.cost_history)
plt.title('Cost History During Training')
plt.xlabel('Iterations')
plt.ylabel('Cost')
plt.grid(True, alpha=0.3)

# Plot data and decision boundary
plt.subplot(2, 2, 2)
plt.scatter(X_train[y_train==0], np.zeros(np.sum(y_train==0)), color='blue', marker='o', label='Class 0', s=50, alpha=0.7)
plt.scatter(X_train[y_train==1], np.zeros(np.sum(y_train==1)), color='red', marker='x', label='Class 1', s=50, alpha=0.7)

# Create a range of X values to plot the decision boundary
X_range = np.linspace(min(X_train)[0] - 1, max(X_train)[0] + 1, 1000).reshape(-1, 1)
y_prob_range = model.predict_proba(X_range)

plt.plot(X_range, y_prob_range, color='green', label='Probability Curve')
plt.axhline(y=0.5, color='black', linestyle='--', label='Decision Boundary (0.5)')

# Mark the decision boundary on X-axis
decision_boundary = -model.bias / model.weights[0]
plt.axvline(x=decision_boundary, color='purple', linestyle='--', label=f'Decision Boundary (x={decision_boundary:.2f})')

plt.title('Logistic Regression Decision Boundary')
plt.xlabel('X')
plt.ylabel('Probability / Class')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot confusion matrix
plt.subplot(2, 2, 3)
cm = confusion_matrix(y_test, y_pred_test)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')

# Plot ROC curve
plt.subplot(2, 2, 4)
fpr, tpr, thresholds = roc_curve(y_test, y_prob_test)
roc_auc = auc(fpr, tpr)
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.title('Receiver Operating Characteristic')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc="lower right")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Evaluate different probability thresholds
thresholds_to_try = np.linspace(0, 1, 11)
threshold_accuracies = []

for threshold in thresholds_to_try:
    y_pred = (y_prob_test >= threshold).astype(int)
    accuracy = accuracy_score(y_test, y_pred)
    threshold_accuracies.append(accuracy)
    print(f"Threshold: {threshold:.1f}, Accuracy: {accuracy:.4f}")

# Plot threshold vs accuracy
plt.figure(figsize=(10, 6))
plt.plot(thresholds_to_try, threshold_accuracies, marker='o')
plt.title('Accuracy vs. Probability Threshold')
plt.xlabel('Threshold')
plt.ylabel('Accuracy')
plt.grid(True)
plt.show()
```

## 8. Naïve Bayesian Classifier from CSV

In this section, we'll implement a Naïve Bayesian Classifier using a sample dataset from a CSV file and compute its accuracy with test data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import seaborn as sns

# Generate a sample CSV file for demonstration purposes
np.random.seed(42)

# Define feature parameters for each class
# We'll create a dataset with 2 features and 2 classes
n_samples_per_class = 100

# Class 0: Normal distributions with different means
class0_feature1 = np.random.normal(0, 1, n_samples_per_class)
class0_feature2 = np.random.normal(1, 1, n_samples_per_class)

# Class 1: Normal distributions with different means
class1_feature1 = np.random.normal(2, 1, n_samples_per_class)
class1_feature2 = np.random.normal(0, 1, n_samples_per_class)

# Combine features and create labels
X = np.vstack([np.column_stack((class0_feature1, class0_feature2)), 
               np.column_stack((class1_feature1, class1_feature2))])
y = np.hstack([np.zeros(n_samples_per_class), np.ones(n_samples_per_class)])

# Create a dataframe with the features and target
data = pd.DataFrame({
    'feature1': X[:, 0], 
    'feature2': X[:, 1],
    'class': y
})

# Save to CSV
csv_path = 'naive_bayes_data.csv'
data.to_csv(csv_path, index=False)
print(f"Created sample CSV file: {csv_path}")
print(data.head())

# Read the data from CSV
data = pd.read_csv(csv_path)
print("\nData loaded from CSV:")
print(data.head())

# Split into features and target
X = data[['feature1', 'feature2']].values
y = data['class'].values

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Implement Naive Bayes classifier from scratch
class NaiveBayes:
    def __init__(self):
        self.classes = None
        self.mean = None
        self.var = None
        self.priors = None
        
    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.classes = np.unique(y)
        n_classes = len(self.classes)
        
        # Initialize mean, variance, and priors
        self.mean = np.zeros((n_classes, n_features))
        self.var = np.zeros((n_classes, n_features))
        self.priors = np.zeros(n_classes)
        
        # Calculate mean, variance, and priors for each class
        for i, c in enumerate(self.classes):
            X_c = X[y == c]
            self.mean[i, :] = X_c.mean(axis=0)
            self.var[i, :] = X_c.var(axis=0) + 1e-10  # Add small value to avoid division by zero
            self.priors[i] = X_c.shape[0] / n_samples
    
    def _pdf(self, class_idx, x):
        # Gaussian probability density function
        mean = self.mean[class_idx]
        var = self.var[class_idx]
        
        # Calculate the exponent
        numerator = np.exp(- (x - mean) ** 2 / (2 * var))
        denominator = np.sqrt(2 * np.pi * var)
        
        return numerator / denominator
    
    def predict(self, X):
        n_samples = X.shape[0]
        n_classes = len(self.classes)
        
        # Calculate posterior probability for each class
        posteriors = np.zeros((n_samples, n_classes))
        
        for i, c in enumerate(self.classes):
            prior = np.log(self.priors[i])
            class_conditional = np.sum(np.log(self._pdf(i, X)), axis=1)
            posteriors[:, i] = prior + class_conditional
        
        # Return class with highest posterior probability
        return self.classes[np.argmax(posteriors, axis=1)]

# Train the model
model = NaiveBayes()
model.fit(X_train, y_train)

# Make predictions
y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)

# Calculate metrics
train_accuracy = accuracy_score(y_train, y_pred_train)
test_accuracy = accuracy_score(y_test, y_pred_test)

# Print results
print(f"\nNaive Bayes Results:")
print(f"Training Accuracy: {train_accuracy:.4f}")
print(f"Testing Accuracy: {test_accuracy:.4f}")
print("\nClass priors:")
for i, c in enumerate(model.classes):
    print(f"Class {c}: {model.priors[i]:.4f}")

print("\nClass means:")
for i, c in enumerate(model.classes):
    print(f"Class {c}: {model.mean[i]}")

print("\nClass variances:")
for i, c in enumerate(model.classes):
    print(f"Class {c}: {model.var[i]}")

# Print classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred_test))

# Visualize the results
plt.figure(figsize=(16, 12))

# Plot training data
plt.subplot(2, 2, 1)
for i, c in enumerate(model.classes):
    plt.scatter(X_train[y_train == c, 0], X_train[y_train == c, 1], 
                label=f'Class {int(c)}', alpha=0.5)
plt.title('Training Data')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot testing data with predictions
plt.subplot(2, 2, 2)
for i, c in enumerate(model.classes):
    plt.scatter(X_test[y_test == c, 0], X_test[y_test == c, 1], 
                marker='o', label=f'True Class {int(c)}', alpha=0.5)

# Highlight misclassified points
misclassified = X_test[y_test != y_pred_test]
plt.scatter(misclassified[:, 0], misclassified[:, 1], 
            marker='x', color='red', s=100, label='Misclassified', alpha=0.8)

plt.title('Test Data and Predictions')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot the decision boundary
plt.subplot(2, 2, 3)

# Create a mesh grid
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                     np.arange(y_min, y_max, 0.1))

# Predict for each point in the mesh
Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plot the decision boundary
plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.RdBu)

# Plot the training points
for i, c in enumerate(model.classes):
    plt.scatter(X_train[y_train == c, 0], X_train[y_train == c, 1], 
                label=f'Class {int(c)}', alpha=0.5)

plt.title('Naive Bayes Decision Boundary')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot confusion matrix
plt.subplot(2, 2, 4)
cm = confusion_matrix(y_test, y_pred_test)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')

plt.tight_layout()
plt.show()

# Visualize the probability distributions for each feature and class
plt.figure(figsize=(16, 6))

# Feature 1 distributions
plt.subplot(1, 2, 1)
x = np.linspace(X[:, 0].min() - 1, X[:, 0].max() + 1, 1000)
for i, c in enumerate(model.classes):
    plt.plot(x, model._pdf(i, np.column_stack((x, np.zeros_like(x))))[:, 0], 
             label=f'Class {int(c)}', linewidth=2)
plt.title('Feature 1 Probability Distributions by Class')
plt.xlabel('Feature 1')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True, alpha=0.3)

# Feature 2 distributions
plt.subplot(1, 2, 2)
x = np.linspace(X[:, 1].min() - 1, X[:, 1].max() + 1, 1000)
for i, c in enumerate(model.classes):
    plt.plot(x, model._pdf(i, np.column_stack((np.zeros_like(x), x)))[:, 1], 
             label=f'Class {int(c)}', linewidth=2)
plt.title('Feature 2 Probability Distributions by Class')
plt.xlabel('Feature 2')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
