# Big Data Sets and Data Distribution

In this notebook, we'll explore working with larger datasets, understand different types of data distributions, and create visualizations with scatter plots.

```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing, load_iris
from scipy import stats

# Set up plotting style
plt.style.use('seaborn')
%matplotlib inline
```

## Part 1: Obtaining Big Data Sets

Let's explore different methods to obtain larger datasets for analysis:

```python
# Method 1: Using built-in datasets from scikit-learn
housing = fetch_california_housing()
housing_df = pd.DataFrame(housing.data, columns=housing.feature_names)
housing_df['PRICE'] = housing.target

print(f"California Housing Dataset Shape: {housing_df.shape}")
print("\nFirst 5 rows of California Housing Dataset:")
housing_df.head()
```

```python
# Method 2: Loading datasets from CSV files (or creating sample data)
sample_data = pd.DataFrame({
    'feature1': np.random.rand(100),
    'feature2': np.random.rand(100),
    'feature3': np.random.rand(100),
    'target': np.random.randint(0, 2, 100)
})

# Save the sample data to CSV
sample_data.to_csv('sample_data.csv', index=False)

# Load the data back from CSV
loaded_data = pd.read_csv('sample_data.csv')
print(f"Loaded Data Shape: {loaded_data.shape}")
print("\nFirst 5 rows of Loaded Data:")
loaded_data.head()
```

## Part 2: Normal Data Distribution

Let's explore the normal distribution and its properties:

```python
# Generate normal distribution data
n_samples = 1000
normal_data = np.random.normal(loc=0, scale=1, size=n_samples)

# Plot histogram of normal distribution
plt.figure(figsize=(10, 6))
plt.hist(normal_data, bins=30, alpha=0.7, color='blue')
plt.title('Normal Distribution Histogram', fontsize=14)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

# Calculate and print statistics
mean = np.mean(normal_data)
median = np.median(normal_data)
std_dev = np.std(normal_data)
skewness = stats.skew(normal_data)
kurtosis = stats.kurtosis(normal_data)

print("Normal Distribution Statistics:")
print(f"Mean: {mean:.4f}")
print(f"Median: {median:.4f}")
print(f"Standard Deviation: {std_dev:.4f}")
print(f"Skewness: {skewness:.4f}")
print(f"Kurtosis: {kurtosis:.4f}")
```

```python
# Normal distribution with different parameters
plt.figure(figsize=(10, 6))

x = np.linspace(-5, 5, 1000)
plt.plot(x, stats.norm.pdf(x, loc=0, scale=1), 'r-', label='μ=0, σ=1')
plt.plot(x, stats.norm.pdf(x, loc=0, scale=0.5), 'g-', label='μ=0, σ=0.5')
plt.plot(x, stats.norm.pdf(x, loc=1, scale=1), 'b-', label='μ=1, σ=1')

plt.title('Normal Distribution PDF with Different Parameters', fontsize=14)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## Part 3: Random Data Distribution

Let's generate and visualize different types of random distributions:

```python
# Generate data from different distributions
n_samples = 1000
uniform_data = np.random.uniform(low=-3, high=3, size=n_samples)
exponential_data = np.random.exponential(scale=1.0, size=n_samples)

# Create subplots for different distributions
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Uniform distribution
axes[0].hist(uniform_data, bins=30, alpha=0.7, color='green')
axes[0].set_title('Uniform Distribution', fontsize=14)
axes[0].grid(True, alpha=0.3)
axes[0].set_xlabel('Value')
axes[0].set_ylabel('Frequency')

# Exponential distribution
axes[1].hist(exponential_data, bins=30, alpha=0.7, color='red')
axes[1].set_title('Exponential Distribution', fontsize=14)
axes[1].grid(True, alpha=0.3)
axes[1].set_xlabel('Value')

plt.tight_layout()
plt.show()

# Print statistics for each distribution
print("Uniform Distribution Statistics:")
print(f"Mean: {np.mean(uniform_data):.4f}")
print(f"Std Dev: {np.std(uniform_data):.4f}")
print(f"Min: {np.min(uniform_data):.4f}, Max: {np.max(uniform_data):.4f}")

print("\nExponential Distribution Statistics:")
print(f"Mean: {np.mean(exponential_data):.4f}")
print(f"Std Dev: {np.std(exponential_data):.4f}")
print(f"Min: {np.min(exponential_data):.4f}, Max: {np.max(exponential_data):.4f}")
```

## Part 4: Scatter Plots for Data Visualization

Let's create scatter plots to visualize relationships between variables:

```python
# Load the Iris dataset for scatter plot examples
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)

print("Iris Dataset Shape:", iris_df.shape)
print("\nIris Dataset First 5 rows:")
iris_df.head()
```

```python
# Basic scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(iris_df['sepal length (cm)'], iris_df['sepal width (cm)'])
plt.title('Sepal Length vs Sepal Width', fontsize=14)
plt.xlabel('Sepal Length (cm)', fontsize=12)
plt.ylabel('Sepal Width (cm)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Color-coded scatter plot by species
plt.figure(figsize=(10, 6))

# Get unique species
species_names = iris_df['species'].unique()
colors = ['blue', 'green', 'red']

# Plot each species with different color
for i, species in enumerate(species_names):
    subset = iris_df[iris_df['species'] == species]
    plt.scatter(subset['petal length (cm)'], subset['petal width (cm)'], 
                c=colors[i], label=species, alpha=0.7)

plt.title('Petal Length vs Petal Width by Species', fontsize=14)
plt.xlabel('Petal Length (cm)', fontsize=12)
plt.ylabel('Petal Width (cm)', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Correlation matrix as heatmap
plt.figure(figsize=(8, 6))
corr_matrix = iris_df.iloc[:, :-1].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Iris Features', fontsize=14)
plt.show()
```

```python
# Advanced scatter plot matrix with Seaborn
sns.pairplot(iris_df, hue='species', height=2)
plt.suptitle('Pair Plot of Iris Dataset Features by Species', y=1.02, fontsize=14)
plt.show()
```
